# NB01 - Import de données avec UI

## 1. Import d'un fichier dans un volume

Dans le dossier Resources/NB01 il un fichier `test.csv` qui peut être utilisé pour cette démo.
Pour le télécharger aller dans le dossier et cliquer sur les trois petits points à côté du fichier et cliquer **Download** :

![image_1790777365118.png](./Resources/image_1790777365118.png "image_1790777365118.png")


Une fois le fichier téléchargé, dans la page **Catalog** naviguer jusqu'au volume créé : `labuser_<username>.sql_analytics.source_files`
et cliquer en haut à droite sur **Upload to Volume** : 

![image_1790778571894.png](./Resources/image_1790778571894.png "image_1790778571894.png")


Et finalement dans la popup renseigner le fichier a importer : 

![image_1790778633096.png](./Resources/image_1790778633096.png "image_1790778633096.png")


## 2. Créer un table à partir d'un fichier CSV avec l'UI

Dans le schéma cliquer en haut à droite sur **Create -> Table** :

![image_1790779689337.png](./Resources/image_1790779689337.png "image_1790779689337.png")

Dans la nouvelle fenêtre ajouter le fichier a importer et si l'affichage de fichier ne se fait pas corrêctement, colonnes pas détéctées, ... changer les paramètres dans **Advanced Details** : 

![image_1790779881081.png](./Resources/image_1790779881081.png "image_1790779881081.png")
