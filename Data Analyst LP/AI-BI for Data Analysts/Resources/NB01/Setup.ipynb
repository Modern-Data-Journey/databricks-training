```python
from databricks.sdk import WorkspaceClient

w = WorkspaceClient()

catalog_name = "labuser_" + spark.sql("SELECT current_user()").collect()[0][0].split("@")[0]
schema_name = "aibi_dashboards"
volume_name = "source_files"

def catalog_exists(name):
    try:
        w.catalogs.get(name)
        return True
    except Exception:
        print("the lab is not set please run the setup cell in the notebook NB00 - Setup et Exploration")
        return False

def schema_exists(catalog, name):
    try:
        w.schemas.get(f"{catalog}.{name}")
        return True
    except Exception:
        print("the lab is not set please run the setup cell in the notebook NB00 - Setup et Exploration")
        return False

def volume_exists(catalog, schema, name):
    try:
        w.volumes.read(f"{catalog}.{schema}.{name}")
        return True
    except Exception:
        print("the lab is not set please run the setup cell in the notebook NB00 - Setup et Exploration")
        return False

def tables_exists(catalog, schema, name):
    try:
        w.tables.get(f"{catalog}.{schema}.{name}")
        return True
    except Exception:
        print("the lab is not set please run the setup cell in the notebook NB00 - Setup et Exploration")
        return False

print(f"Catalog '{catalog_name}' exists: {catalog_exists(catalog_name)}")
print(f"Schema '{catalog_name}.{schema_name}' exists: {schema_exists(catalog_name, schema_name)}")
print(f"Volume '{catalog_name}.{schema_name}.{volume_name}' exists: {volume_exists(catalog_name, schema_name, volume_name)}")

tables_to_check = [
    "ca_customers",
    "ca_opportunities",
    "ca_orders",
    "ca_products",
    "ca_sales_reps",
    "ca_sales_targets",
]

print()
for table in tables_to_check:
    print(f"Table '{catalog_name}.{schema_name}.{table}' exists: {tables_exists(catalog_name, schema_name, table)}")
```
