## AI/BI Dashboard Architecture

### Description des couches

| Couche | Rôle | Détail |
|--------|------|--------|
| **Sources** | Fichiers `.csv` bruts | Données d'origine à ingestér |
| **Ingestion** | Notebook `NB00 / Setup` | Copie les fichiers dans le volume `labuser_<username>.aibi_dashboards.source_files` puis crée les tables UC |
| **Unity Catalog** | Tables du schéma `aibi_dashboards` | `ca_orders`, `ca_customers`, `ca_products`, `ca_sales_reps`, `ca_opportunities`, `ca_sales_targets` |
| **Dataset** | Requêtes SQL du dashboard | Jointures, agrégations et calculs de KPIs alimentant les widgets |
| **Dashboard** | AI/BI Dashboard | Visualisations interactives : KPIs, graphiques, tables et filtres |
